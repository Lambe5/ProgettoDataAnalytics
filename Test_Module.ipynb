```python
import pandas as pd

def load_df(path1, path2, path3):
    df1 = pd.read_csv(path1)
    df2 = pd.read_csv(path2)
    df3 = pd.read_csv(path3)
    return (df1, df2, df3)
```

```python
(tabNet_pca, tabNet_scaled_no_outliers, tabNet_scaled) = load_df(
    "df_Res_tabnet_PCA.csv", "df_Res_tabnet_Scaled_no_outliers.csv", "df_Res_tabnet_Scaled.csv")
```

```python
tabNet_pca.sort_values(by='test_mean_absolute_error', ascending=True)
```
